```python
import pandas as pd
data = {'color':['red','blue','orange','black','Yellow']}
df = pd.DataFrame(data)
df
```

```python
#Perfoming one-hot encoding 
one_hot_encoding = pd.get_dummies(df['color'],prefix='color')
one_hot_encoding
```

```python
#combinig the enconded data with original data
df = pd.concat([df,one_hot_encoding],axis=1)
df
```

```python
df = df.drop('color',axis=1)
df
```
