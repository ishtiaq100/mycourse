# Linear Regration implementation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
from sklearn.datasets import fetch_openml
boston = fetch_openml(name='boston',version=1,as_frame=True,parser='pandas')
```

```python
boston
```

```python
#convert into df
df= boston.frame
df.head()
```

```python
#create the target column
df['price'] = boston.target
df.head()
```

```python
#divide the dataset into dependent and independent features
```

```python
x = df.iloc[:,:-1]
x
```

```python
y = df.iloc[:,-1]
y
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
x = x.astype(float)
y = y.astype(float)

lin_reg = LinearRegression()
mse = cross_val_score(lin_reg,x,y,scoring='neg_mean_squared_error',cv=5)
print(mse)
```

```python
mean_mse = np.mean(mse)
print("mse scores per fold ",mse)

print("Avg negitive mse  ",mean_mse)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
lin_reg.fit(x_train,y_train)
```

```python
predection = lin_reg.predict(x_test)
predection
```

```python
#compare a few actual values vs predicted values
comparision_df = pd.DataFrame({'Actual':y_test,'Prdicted':predection})
comparision_df.head()
```

```python
from sklearn import metrics
#1 R-Squered socore (best possible score is 1.0)
r2 = metrics.r2_score(y_test,predection)
mae = metrics.mean_absolute_error(y_test,predection)
rmse = np.sqrt(metrics.mean_squared_error(y_test,predection))
print(f"R Square score : {r2:.4f}")
print(f"Mean absolute error : {mae:.4f}")
print(f"Root Mean Square error : {rmse:.4f}")
```

# Ridge Regration

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
ridge = Ridge()
```

```python
params = {'alpha':[1e-15,1e-10,1e-8,1e-2,1,5,10,20]}
rigdge_regrestion = GridSearchCV(ridge,params,scoring='neg_mean_squared_error',cv=5)
rigdge_regrestion.fit(x,y)
```

# Lasso Regression

```python
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV
lasso = Lasso()
params = {'alpha':[1e-15,1e-10,1e-8,1e-2,1,5,10,20]}
lasso_regrestion = GridSearchCV(lasso,params,scoring='neg_mean_squared_error',cv=5)
lasso_regrestion.fit(x,y)
```

```python
print(lasso_regrestion.best_params_)
print(lasso_regrestion.best_score_)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
# Ensure data is numeric
x = x.astype(float)
y = y.astype(float)
# scale the feature for (Essentials for Ridge and lasoo)
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
x_train,x_test,y_train,y_test = train_test_split(x_scaled,y,test_size=0.2,random_state=42)
```

# Linear Regression

```python
from sklearn.linear_model import LinearRegression
#intlize and train
lin_reg = LinearRegression()
lin_reg.fit(x_train,y_train)
# predirect and calculate accurancy
lin_pred = lin_reg.predict(x_test)
lin_acc = r2_score(y_test,lin_pred)*100
print(f"Linear Regrestion Accuracy: {lin_acc:.2f}%")
```

# Ridge Regression

```python
from sklearn.linear_model import Ridge
ridge_reg = Ridge(alpha=10.0)
ridge_reg.fit(x_train,y_train)

#predict and calculate accurracy
ridge_pred = ridge_reg.predict(x_test)
ridge_acc = r2_score(y_test,ridge_pred)*100
print(f"Ridge Regression Accuracy : {ridge_acc:.2f}%")
```
